```python
# Histograms
plt.figure(figsize=(8, 5))
sns.histplot(df["Age"].dropna(), bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Fare distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["Fare"], bins=30, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Passenger class distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["Pclass"], bins=3, discrete=True)
plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Create Boxplots
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Age"])
plt.title("Age Boxplot")
plt.xlabel("Age")
plt.show()
```

```python
# Fare boxplot
plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Fare"])

plt.title("Fare Boxplot")
plt.xlabel("Fare")

plt.show()
```

```python
# SibSp boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["SibSp"])
plt.title("SibSp Boxplot")
plt.xlabel("Number of Siblings/Spouses")
plt.show()
```

```python
# Parch boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x=df["Parch"])
plt.title("Parch Boxplot")
plt.xlabel("Number of Parents/Children")
plt.show()
```

```python
print(df["Sex"].value_counts())
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sex")
plt.title("Passenger Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Analyze passenger class
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Pclass")
plt.title("Passenger Distribution by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Analyze survival
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Survived")
plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Survival by gender
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Survival by Gender")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Survival by passenger class
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Pclass", hue="Survived")
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Pairplot
pairplot_data = df[["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]].dropna()
sns.pairplot(pairplot_data, hue="Survived")
plt.show()
```

```python
# Claculate Correlation Matrix
correlation_matrix = df.select_dtypes(include=np.number).corr()
print(correlation_matrix)
```

```python
# visualize this Correlation Matrix
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Matrix of Titanic Dataset")
plt.show()
```

```python
# Calculate skewness
print("Skewness of numerical features:")
print(df.select_dtypes(include=np.number).skew())
```

```python
# specifically check
print("Age skewness:", df["Age"].skew())
print("Fare skewness:", df["Fare"].skew())
```

```python
# Identify anomalies/outliers
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
# count the possible outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]
print("Number of Fare outliers:", len(fare_outliers))
```
